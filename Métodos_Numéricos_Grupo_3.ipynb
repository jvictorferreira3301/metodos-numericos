### 1. Varredura de possíveis raízes
* Implementar busca incremental de possíveis raízes num intervalo de interesse, com base na seção 5.4, página 113.
* Receber do usuário a expressão da função, o intervalo de interesse e o número N de subdivisões.
* Calcular o sinal da função para cada subdivisão.
* Retornar os subintervalos com mudança de sinal ou zero da função.
* Retornar gráfico mostrando todas as subdivisões e destacando as detecções de mudança de sinal.
* Solicitar ao usuário se deseja refinar N ou proceder para o cálculo.

```python
import numpy as np
import matplotlib.pyplot as plt

intervalo_1 = float(input("Insira a primeira extremidade do intervalo: ").strip)
intervalo_2 = float(input("Insira a segunda extremidade do intervalo: ").strip)

# função
def f(x):
    return 5*x**2 + 3*x + 2

# domínio do gráfico
x = np.linspace(intervalo_1, intervalo_2, 200)
y = f(x)

plt.plot(x, y)

# intervalo que você quer destacar
a = -2
b = 1

# linhas verticais nas extremidades
plt.axvline(x=a, linestyle='--')
plt.axvline(x=b, linestyle='--')

# opcional: marcar os pontos também
plt.scatter([a, b], [f(a), f(b)])

plt.title('Intervalo destacado em f(x) = 5x² + 3x + 2')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()

plt.show()
```
